```python
import open3d as o3d
import numpy as np
import matplotlib.pyplot as plt

def visualize_o3d(xyz, labels=None, rgb=None, point_size=1.0):
    """
    xyz: torch.Tensor (N,3)
    labels: torch.Tensor (N,)
    rgb: torch.Tensor (N,3) floating 0..1
    """

    xyz = xyz.detach().cpu().numpy()

    pcd = o3d.geometry.PointCloud()
    pcd.points = o3d.utility.Vector3dVector(xyz)

    if rgb is not None:
        rgb = rgb.detach().cpu().numpy()
        rgb = rgb.astype(np.float32)
        if rgb.max() > 1.0:
            rgb = rgb / 255.0

        pcd.colors = o3d.utility.Vector3dVector(rgb)


    elif labels is not None:
        labels = labels.detach().cpu().numpy()
        # colormap
        cmap = plt.get_cmap("tab20")
        max_label = max(labels.max(), 1)
        colors = cmap(labels / max_label)[:, :3]
        pcd.colors = o3d.utility.Vector3dVector(colors)

    # Включаем красивое отображение
    vis = o3d.visualization.Visualizer()
    vis.create_window(width=1000, height=800)
    vis.add_geometry(pcd)

    opt = vis.get_render_option()
    opt.point_size = point_size
    opt.background_color = np.asarray([0, 0, 0])

    vis.run()
    vis.destroy_window()
```
